```python
# 1.Program to calculate number of seconds in a given number of days
days = int(input("Enter number of days to convert to seconds: "))

# formula is 1 day = 24hrs * 3600 seconds/hr
seconds = days * 24 * 3600

print(f"The number of seconds in {days} day(s) is {seconds}\n")
```

```python
# 2. Program to calculate the volume of a sphere of a given radius
radius = int(input("Enter the radius of the sphere: "))

# we need to import the math library to access pi constant
import math

# volume of a sphere is 4/3 * pi * r**3
volume = 4/3 * math.pi * radius ** 3
print(f"The volume of the sphere is {volume}\n")
```

```python
# 3. Program to calculate area and perimeter of a square
length = int(input("Enter the length of the square: "))

# Function to calculate area = Length * length
def area(length):
  return length ** 2

# Function to calculate perimeter = length * 4
def perimeter(length):
  return length * 4

print(f"The area is {area(length)} and the perimeter is {perimeter(length)} ")

```

```python
# 4. Program to determine whether a character is upper or lower case
# Upper function
def upper(ch):
  return ch.isupper()

def lower(ch):
  return ch.islower()

character  = input("Enter character to evaluate case: ")
if upper(character):
  print("The character is in upper case ")
elif lower(character):
  print("The character is in lower case ")
```

```python
# 5. Program for pseudo code equivalent

# Variable init
x, y = 0, 20
while y >= 6:
  # reduce y by 4 every iteration
  y -= 4
  # increase x by 2/y every iteration
  x += 2/y

# Display x
print(x)
```

```python
# 6. Program to populate an array and find the average of the values

# initialize the array
myArr = []

# loop to populate the array
for i in range(5):
  element = int(input("Enter an element to populate the array: "))
  myArr.append(element)

# Calculate the average of the elements = total / 5
average = sum(myArr) / 5
print(average)
```
